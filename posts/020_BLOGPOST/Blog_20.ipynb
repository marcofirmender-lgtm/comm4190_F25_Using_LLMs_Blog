
In the last two sections of Part I, Ethan Mollick deepens his reflection on the strange partnership between humans and machines. After introducing us to the idea of “creating alien minds,” he turns toward the question of *alignment* — how we can ensure that these new intelligences not only work for us, but also *with* us. Sections 2 and 3, “Aligning the Alien” and “Four Rules for Co-Intelligence,” read as both philosophical guidance and a practical roadmap for living productively alongside AI.

In **“Aligning the Alien,”** Mollick explores the difficulty of teaching something that doesn’t think like a human to share human values. Large language models, he explains, are not conscious moral agents; they don’t understand empathy, truth, or purpose. Yet, through training and reinforcement, we attempt to align their behaviors with what we *want* them to represent — safety, helpfulness, creativity, and accuracy. Mollick highlights the paradox of this effort: we’re trying to make alien systems “human enough” to be trusted, even though they fundamentally operate on pattern recognition, not ethical judgment. He compares this process to raising a precocious but inscrutable child — capable of brilliance, yet always at risk of misunderstanding us.

Then, in **“Four Rules for Co-Intelligence,”** Mollick distills his guidance into a set of principles for effective collaboration with AI. While each rule centers on different aspects of human–machine partnership, together they emphasize a mindset of *active engagement* rather than blind dependence. He urges readers to (1) treat AI as a colleague, not a tool, (2) be clear and intentional in communication, (3) verify and refine AI’s output through human judgment, and (4) continually learn from the interaction. These rules reflect his belief that successful co-intelligence requires curiosity, skepticism, and a willingness to experiment.

Ultimately, Mollick’s argument across both sections is that the challenge of AI isn’t just technical — it’s cultural and moral. Aligning the alien is about aligning *ourselves* as well: defining what kind of intelligence we value, and how we choose to amplify it. His four rules serve as a compass for that journey, reminding us that the future of intelligence will be *shared* — a partnership between human insight and machine possibility.
