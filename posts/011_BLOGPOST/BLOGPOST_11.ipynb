OpenAI and AMD have announced a major multi-year partnership that marks a significant shift in the AI hardware landscape. The deal, unveiled on October 6, 2025, makes AMD a core compute supplier for OpenAI’s growing infrastructure needs. Under the agreement, OpenAI plans to deploy up to **6 gigawatts** of AMD’s next-generation GPUs, beginning with a 1-gigawatt rollout in 2026 using the new **Instinct MI450** chips. In exchange, AMD has granted OpenAI a **warrant to purchase up to 160 million AMD shares**—roughly 10% of the company—at a nominal price. These shares will vest over time as OpenAI meets deployment milestones and AMD achieves certain performance and stock price targets. AMD expects the partnership to generate **tens of billions of dollars in revenue** over the coming years.

Strategically, the deal allows **OpenAI to diversify beyond Nvidia**, securing more compute power and reducing dependence on a single supplier as it scales its AI models. For AMD, this partnership represents both a validation and a challenge—a chance to prove its GPUs can rival Nvidia’s dominance in AI computing. The equity structure aligns both companies’ incentives, effectively making OpenAI a long-term stakeholder in AMD’s success. Still, the agreement comes with risks: the success of AMD’s chips will depend on their ability to deliver competitive performance at massive scale, and the infrastructure demands—datacenters, energy, and cooling—will be enormous. If successful, however, the partnership could **reshape the balance of power in the AI hardware industry**, accelerating innovation, increasing competition, and lowering costs across the ecosystem.
