## 📅 Phased Training Plan (Starting January 2025)

Since you have 17 months, we can divide the training into four main phases. This provides a slow, safe, and sustainable buildup.

### 1. Base Building Phase (January – May 2025: 5 months)

The goal here is to establish a solid running foundation and make running a consistent habit without burning out.

* **Focus:** Easy, comfortable runs, focusing on consistency and form.
* **Weekly Mileage:** Start low (e.g., 10-15 miles per week) and increase by no more than $10\%$ per week.
* **Key Runs:**
    * **3-4 Easy Runs** (30-45 minutes) on your flexible weeknight slot (6:00 PM).
    * **1 Long Run** on Sunday (Start at 3 miles and increase by 1 mile every 1-2 weeks).

### 2. Strength & Endurance Phase (June – September 2025: 4 months)

Once your body is accustomed to running 4-5 times a week, we can introduce more challenging workouts.

* **Focus:** Increasing weekly mileage and adding intensity to build speed and stamina.
* **Weekly Mileage:** Continue the gradual increase, aiming to comfortably reach 25-30 miles per week.
* **Key Runs:**
    * **Easy Runs:** Maintain 3-4 easy runs.
    * **Tempo/Speed Run:** Replace one easy run with a challenging session (e.g., Tuesday at 6:00 PM). This could be a tempo run (sustained hard effort) or an interval session.
    * **Long Run:** Continue increasing the Sunday run, aiming to reach 8-10 miles comfortably by the end of this phase.

### 3. Peak Training Phase (October 2025 – February 2026: 5 months)

This is the core of your marathon training, where the biggest mileage is logged.

* **Focus:** Building the necessary endurance to complete the marathon distance.
* **Weekly Mileage:** This is where you'll hit your peak (up to 40-55 miles, depending on your goal). Crucially, incorporate "down weeks" (lower mileage) every 3-4 weeks to allow your body to recover.
* **Key Runs:**
    * **Long Run:** The Sunday run should progress to **12, 14, 16, 18, and eventually 20 miles**. *Only run 20 miles once or twice.*
    * **Mid-Week Medium Run:** A slightly longer run (6-8 miles) on a Wednesday or Thursday.
    * **Quality Run:** Continue the Tuesday tempo/speed work.

### 4. Taper Phase (March – May 2026: 2 months)

This is the final countdown before the race.

* **Focus:** Resting your legs and storing energy. The hard work is done!
* **Weekly Mileage:** Significantly reduce mileage (e.g., $75\%$ of peak, then $50\%$, then $25\%$).
* **Key Runs:** Runs are much shorter, focusing on feeling refreshed. Keep the Tuesday speed session short and fast to maintain muscle memory, but stop the very long runs. The final week is mostly rest.
