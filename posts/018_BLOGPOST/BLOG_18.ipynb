
Ethan Mollick’s Co-Intelligence opens with a vivid and restless energy in its introduction, aptly titled “Sleepless Nights.” This section captures both the awe and anxiety surrounding artificial intelligence’s rapid rise, grounding Mollick’s reflections in personal experience and the intellectual hum of late-night curiosity. He writes not as a detached scholar but as a professor who, like so many of us, finds his mind racing with questions about what AI means for creativity, work, and the human spirit.

Mollick begins by describing his sleeplessness as the product of encountering something profoundly new — a technology that seems to think with us rather than for us. He recounts moments with his students when ideas would “snowball,” sparked by the interaction between human imagination and generative AI. This snowballing — where one suggestion leads to another, accelerating the creative process — reveals the essence of co-intelligence. It’s not about replacing human thought, but rather expanding its reach, allowing students to push boundaries faster than ever before. Mollick’s classroom becomes a laboratory of discovery, where the line between tool and collaborator begins to blur.

A key theme he introduces is the notion of AI as a ** general purpose technology** — something that reshapes not just how we do things, but why we do them. Unlike narrow innovations such as the steam engine or the light bulb, AI has the potential to touch every domain of human activity, redefining what “purpose” means in work, art, and knowledge creation. Mollick treats this with both excitement and humility, acknowledging that humanity is only beginning to understand the implications.

Finally, Mollick surveys the extraordinary surge in productivity following the debut of ChatGPT. From coding to marketing to education, industries witnessed exponential gains in output and efficiency — a phenomenon he sees as both thrilling and disorienting. The “sleepless nights” become a metaphor for our collective moment: a world staying up late to glimpse its future, trying to balance wonder with responsibility as we step into the era of co-intelligence.
